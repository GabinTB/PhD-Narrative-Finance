# Official `ravenpack-api`

```python
import os
from pathlib import Path
from dotenv import load_dotenv; load_dotenv()

from ravenpackapi import RPApi

api = RPApi(
    # product="rpa",
    product="edge",
    api_key=os.environ.get("RAVENPACK_API_KEY")
)
```

```python
universe = [
    "RavenPack",
    {'ticker': 'AAPL'},
    {  # Amazon, specifying various fields
        "client_id": "12345-A",
        "date": "2017-01-01",
        "name": "Amazon Inc.",
        "entity_type": "COMP",
        "isin": "US0231351067",
        "cusip": "023135106",
        "sedol": "B58WM62",
        "listing": "XNAS:AMZN"
    },

]
mapping = api.get_entity_mapping(universe)

# in this case we match everything
assert len(mapping.matched) == len(universe)
assert [m.name for m in mapping.matched] == [
    "RavenPack Ltd.",
    "Apple Inc.",
    "Amazon.com Inc."
]
```

# Custom `edge_api`

```python
import os
from dotenv import load_dotenv; load_dotenv()

from ravenpack.edge_api import RavenPackClient

rp = RavenPackClient(api_key=os.environ.get("RAVENPACK_API_KEY"))
rp.status()
```

```python
# Taxonomy: browse a couple of event categories
taxonomy = rp.taxonomy.query(categories=["earnings-above-expectations", "product-recall"])
taxonomy.categories
```

```python
# Annotations: storage/usage quota, existing folders and uploaded files.
# Annotations is a separate RavenPack product (upload.ravenpack.com) from Edge data/streaming,
# so this key may not be entitled to it even though `status()` and `taxonomy` above work fine.
from ravenpack.edge_api import AuthenticationError

try:
    quota = rp.annotations.quota()
    print(quota)
except AuthenticationError as e:
    print(f"Annotations API not accessible with this key: {e}")
```

```python
try:
    folders = rp.annotations.folders.list()
    files = rp.annotations.files.list()
    print(folders, files)
except AuthenticationError as e:
    print(f"Annotations API not accessible with this key: {e}")
```
